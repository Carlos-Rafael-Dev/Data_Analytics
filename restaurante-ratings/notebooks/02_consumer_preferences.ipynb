# Conhecendo e tratando a tabela Consumer_preferences.csv #

```python
# Importando o pandas
# Importando a tabela como 'df_consumer_preferences'

import pandas as pd

df_consumer_preferences = pd.read_csv("../data/raw/consumer_preferences.csv")

display("Preferencias dos clientes", df_consumer_preferences)
```

```python
# Visão geral da tabela
# Verificar se tem valores vazios
# Descoberta: a tabela não tem valores vazios

df_consumer_preferences.info()
```

```python
# Confirmar se a coluna "Consumer_ID" tem 138 clientes
# A coluna tem 138 valores únicos, então está correto
# Existem 101 preferencias diferentes

clientes_valores_unicos = df_consumer_preferences["Consumer_ID"].nunique()
preferencias_valores_unicos = df_consumer_preferences["Preferred_Cuisine"].nunique()

print(clientes_valores_unicos, "clientes")
print(preferencias_valores_unicos, "preferencias")
```
