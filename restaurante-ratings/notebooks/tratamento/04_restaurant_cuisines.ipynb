# Conhecendo e tratando a tabela restaurant_cuisines.csv #

```python
# Importando o pandas
# Importando a tabela como 'df_restaurant_cuisines'

import pandas as pd

df_restaurant_cuisines = pd.read_csv("../data/raw/restaurant_cuisines.csv")

display("Restaurant cuisines", df_restaurant_cuisines)
```

```python
# Visão geral da tabela
# Verificar se tem valores vazios
# Descoberta: a tabela não tem valores vazios

df_restaurant_cuisines.info()
```

```python
# Confirmar se a tabela tem valores duplicados
# Existem 17 valores iguais
# Mostrando que um restaurante pode ter mais de uma categoria

restaurants_valores_duplicados = df_restaurant_cuisines["Restaurant_ID"].duplicated().sum()
cuisines_valores_duplicados = df_restaurant_cuisines["Restaurant_ID"].duplicated().sum()

print(restaurants_valores_duplicados, "restaurants")
print(cuisines_valores_duplicados, "cuisines")
```

```python
# Verificar se cada coluna tem outlier
# Conclusao: A tabela não precisa ser tratada

for coluna in df_restaurant_cuisines:
    print(f"\n--- Coluna `{coluna}` ({df_restaurant_cuisines[coluna].count()}) ---")
    display(df_restaurant_cuisines[coluna].unique())
```

```python
# Contamos a frequencia de cada valor unico na coluna
df_restaurant_cuisines['Cuisine'].value_counts()
```

```python
# Exportar a tabela processada
df_restaurant_cuisines.to_csv('restaurant_cuisines_processada.csv', index=False, encoding='utf-8-sig')
```
