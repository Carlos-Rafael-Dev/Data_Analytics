# Conhecendo e tratando a tabela Consumer.csv #

```python
# Importando o pandas
# Importando a tabela como 'df_consumers'

import pandas as pd

df_consumers = pd.read_csv("../data/consumers.csv")

display("Clientes", df_consumers)
```

```python
# Visão geral da tabela
# Verificar se tem valores vazios
# Descoberta: a tabela tem pouco valores vazios

df_consumers.info()
```

```python
# Confirmar se a coluna "Consumer_ID" tem valores únicos
# A coluna tem 138 valores únicos, então está correto

clientes_valores_unicos = df_consumers["Consumer_ID"].nunique()

print(clientes_valores_unicos, "clientes")
```

```python
# Verificar se cada coluna tem outlier
# confimando os dados vazios em algumas colunas "nan"

for coluna in df_consumers:
    print(f"\n--- Coluna `{coluna}` ({df_consumers[coluna].count()}) ---")
    display(df_consumers[coluna].unique())
```

```python
# ---------------   TRATAMENTO  ---------------
# Preencher valores vazios em colunas especificas
# Usando o método Moda (preencher com o valor mais frequente)

df_consumers["Smoker"] = df_consumers["Smoker"].fillna(df_consumers["Smoker"].mode()[0])
df_consumers["Budget"] = df_consumers["Budget"].fillna(df_consumers["Budget"].mode()[0])
df_consumers["Children"] = df_consumers["Children"].fillna(df_consumers["Children"].mode()[0])
df_consumers["Marital_Status"] = df_consumers["Marital_Status"].fillna(df_consumers["Marital_Status"].mode()[0])
df_consumers["Transportation_Method"] = df_consumers["Transportation_Method"].fillna(df_consumers["Transportation_Method"].mode()[0])

# Para confirmar se o preenchimento foi realizado
df_consumers.info()
```

```python
# Salva a tabela tratada
df_consumers.to_csv("consumers_tratada.csv", index=False)
```
