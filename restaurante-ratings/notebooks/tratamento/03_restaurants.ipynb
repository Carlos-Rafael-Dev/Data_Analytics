# Conhecendo e tratando a tabela Restaurants.csv #

```python
# Importando o pandas
# Importando a tabela como 'df_restaurants'

import pandas as pd

df_restaurants = pd.read_csv("../data/raw/restaurants.csv")

display("Restaurants", df_restaurants)
```

```python
# Visão geral da tabela
# Verificar se tem valores vazios
# Descoberta: a tabela não tem valores vazios

df_restaurants.info()
```

```python
# Confirmar se a coluna "Restaurant_ID" tem 130 restaurantes
# A coluna alcohol_service tem 87 valores vazios
# A coluna parking tem 65 valores vazios

restaurants_valores_unicos = df_restaurants["Restaurant_ID"].nunique()
parking_valores_vazios = df_restaurants["Parking"].isna().sum()
alcohol_valores_vazios = df_restaurants["Alcohol_Service"].isna().sum()

print(restaurants_valores_unicos, "restaurants")
print(parking_valores_vazios, "parking")
print(alcohol_valores_vazios, "alcohol service")
```

```python
# Verificar se cada coluna tem outlier
# Conclusao: A tabela precisa preencher valores vazios

for coluna in df_restaurants:
    print(f"\n--- Coluna `{coluna}` ({df_restaurants[coluna].count()}) ---")
    display(df_restaurants[coluna].unique())
```

```python
# As colunas Alcohol_Service e Parking precisam ser tratadas
# Nan não é uma categoria válida, de acordo com o data dictionary
# Alcohol service [nan = no alcohol, wine & beer, full bar]
# Parking [nan = none, yes, public, valet]

df_restaurants['Alcohol_Service'] = df_restaurants['Alcohol_Service'].fillna('no alcohol')
df_restaurants['Parking'] = df_restaurants['Parking'].fillna('none')

display(df_restaurants['Alcohol_Service'].unique())
display(df_restaurants['Parking'].unique())
```

```python
# Para conferir se ainda tem valores nan na tabela
df_restaurants.isna().sum()
```

```python
# Zip Code tem valores nan, mas essa coluna não será usada na analise
# Excluimos essa coluna

df_restaurants.drop(columns=['Zip_Code'])
```

```python
# Exportamos a tabela processada
df_restaurants.to_csv('restaurants_tratada.csv', index=False, encoding='utf-8-sig')
```
