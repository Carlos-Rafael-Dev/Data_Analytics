# Conhecendo e tratando a tabela Ratings.csv #

```python
# Importando o pandas
# Importando a tabela como 'df_ratings'

import pandas as pd

df_ratings = pd.read_csv("../data/raw/ratings.csv")

display("Ratings", df_ratings)
```

```python
# Visão geral da tabela
# Verificar se tem valores vazios
# Descoberta: a tabela não tem valores vazios

df_ratings.info()
```

```python
# Confirmar se a coluna "Restaurant_ID" tem 130 restaurantes
# 138 consumers com valores unicos
# 3 tipos unicos de notas, nas categorias (total, comida, servico)
# 0 = ruim, 1 = bom, 2 = excelente

restaurants_valores_unicos = df_ratings["Restaurant_ID"].nunique()
consumers_valores_unicos = df_ratings["Consumer_ID"].nunique()
overall_valores_unicos = df_ratings["Overall_Rating"].nunique()
food_valores_unicos = df_ratings["Food_Rating"].nunique()
service_valores_unicos = df_ratings["Service_Rating"].nunique()

print(restaurants_valores_unicos, "restaurants")
print(consumers_valores_unicos, "consumers")
print(overall_valores_unicos, "overall")
print(food_valores_unicos, "food")
print(service_valores_unicos, "service")
```

```python
# Verificar se cada coluna tem outlier
# Conclusao: A tabela não precisa ser tratada 

for coluna in df_ratings:
    print(f"\n--- Coluna `{coluna}` ({df_ratings[coluna].count()}) ---")
    display(df_ratings[coluna].unique())
```
